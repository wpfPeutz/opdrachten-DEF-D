##### Antwoord 3b

Voor de oppervlaktelading in het midden van de schijf kun je de wet van Gauss gebruiken. Gaussian surface is een pillendoosje dat door de schijf gaat in het midden. De flux wordt dan E*A. Vanaf hier nemen we aan dat factor van de circulaire platen ongeveer gelijk is aan de factor voor vierkante platen. De factor die de capaciteit van de vlakke plaat approximatie linkt aan de werkelijke capaciteit noemen we vanaf nu de idealiteitsfactor.

````python
# Schrijf hier jouw code die de vlakke plaat capaciteit omschrijft naar de werkelijke capaciteit

import numpy as np  
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit    


#De berekeningen van sanity check 3 van de fysieke ontwerpopdracht:

def functieNaam(oppervlakte, afstand, diëlektrische_constante):
    epsilon_0 = 8.854187817e-12  
    
    # Formule: C = ε0 * εr * A / d
    output = epsilon_0 * diëlektrische_constante * oppervlakte / afstand
    
    return output

oppervlakte_test= 0.01  # in vierkante meters, 10 bij 10 cm
afstand_test = 0.0022  # in meters, 1 mm
diëlektrische_constante_test = 2  # voor papier 3, voor lucht 1. Gewogen met hoe weinig papier er is, schatten wij het op 2
C = [functieNaam(oppervlakte_test, afstand_test, diëlektrische_constante_test)]


print(f"Berekende capaciteit plaatcondensator groep 56 is", C, "Farad")
print(f"Dit komt overeen met {C[0]*1e12:.1f} pF als capaciteit")

Vin= 8 # in Volt
Ceind= 80.5e-12 # in Farad
f= 100000 # in Hz
Rshunt= 1000 # in Ohm


Vout= Vin*2*np.pi*f*Ceind*Rshunt
print(f"De verwachte uitgangsspanning is {Vout:.2f} Volt")

#Berekenen van de werkelijke capaciteit van een vlakke plaat condensator met een oppervlakte van 0.01 m^2, een afstand van 0.0022 m en een relatieve permittiviteit van 2, rekening houdend met rand effecten.

def C_niet_ideaal(S, d, epsilon_r):
    epsilon_0 = 8.854187817e-12
    w = np.sqrt(S) # breedte van de vierkante plaat
    b = d / w      # aspectratio
    
    # Ideale capaciteit (C0)
    C0 = (epsilon_0 * epsilon_r * S) / d
    
    # Correctiefactor (CSN) voor vierkante platen (Nishiyama & Nakamura Eq. 29)
    # Voor b < 0.1 gebruiken we de trendlijn van de studie
    C_sn_factor = 1 + 2.343 * (b**0.891)
    
    actual_C = C0 * C_sn_factor
    return C0, actual_C, b

# Input uit sanity check
S_test = 0.01
d_test = 0.0022
er_test = 2

C0, C_actual, b_val = C_niet_ideaal(S_test, d_test, er_test)

print(f"--- Resultaten Gecorrigeerde Capaciteit ---")
print(f"Aspectratio (b):              {b_val:.4f}")
print(f"Ideale capaciteit (C0):       {C0*1e12:.2f} pF")
print(f"Correctiefactor (Edge effect): {C_actual/C0:.4f}")
print(f"Daadwerkelijke theoretische C: {C_actual*1e12:.2f} pF")


#idealiteitsfactor = C_actual / C0
idealiteitsfactor = C_actual / C0
print(f"De idealiteitsfactor van de condensator is: {idealiteitsfactor:.4f}")


#omrekenen van de theoretische calibaratiegrafiek naar de werkelijke capaciteit:

import numpy as np
import matplotlib.pyplot as plt

# Parameters uit de studie en sanity check
epsilon_0 = 8.854187817e-12
epsilon_r = 1.0  
S = 0.01         
w = 0.1          
z0 = 0.015       # 15 mm startafstand
k = 4.335e-6     # Geschatte samendrukking in m/g

# Massa range
m = np.linspace(0, 3345, 100)
d = z0 - k * m   # De afstand d als functie van massa m

# Berekening met Edge Effect (Nishiyama & Nakamura Eq. 29)
b = d / w
C0 = (epsilon_0 * epsilon_r * S) / d
C_theory = C0 * (1 + 2.343 * (b**0.891))

m1= 973.3
m2=972.2
m3= 938.1
m4= 461.3
M = np.array([0, m4,m1+m4,m1+m3,m1+m2,m1+m3+m4,m1+m2+m3,m1+m2+m3+m4])
C = np.array([150,185,208, 238, 250, 465, 750, 875])*1e-12

def fitfunctie(C, a, b):
    return a-b/C
curve_fit(fitfunctie, C, M)
popt, pcov = curve_fit(fitfunctie, C, M)
a_fit, b_fit = popt
x_fit = np.linspace(np.min(C), np.max(C), 100)
y_fit = fitfunctie(x_fit, a_fit, b_fit)

# Plotten
plt.figure(figsize=(10, 6))
plt.plot(C_theory * 1e12, m, 'r-', label='Theoretisch Model (incl. Edge Effect)')
plt.errorbar(C*1e12, M, yerr=0.5, xerr=0.01*C*1e12, fmt='o', label='Gemeten data')
plt.plot(x_fit*1e12, y_fit, '-', label=f'Fit: m = {a_fit:.2f} * C + {b_fit:.2f}')
plt.xlabel("Capaciteit (pF)")
plt.ylabel("Massa (g)")
plt.title("Theoretische Kalibratiegrafiek: Massa als functie van Capaciteit")
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.show()

print(f"Theoretisch bereik: {min(C_theory*1e12):.2f} pF tot {max(C_theory*1e12):.2f} pF")

````
